```python
import pandas as pd

for column in df_numeric_column:
    skew_value = df[column].skew()
    if -0.5 <= skew_value <= 0.5:
        print(column, "-> Use mean")
    else:
        print(column, "-> Use median")
```

```python
df["age"] = df["age"].fillna(
    df["age"].mean()
)
df["bmi"] = df["bmi"].fillna(
    df["bmi"].mean()
)
df["children"] = df["children"].fillna(
    df["children"].median()
)
```

```python
df["sex"] = df["sex"].fillna(
    df["sex"].mode()[0]
)
df["smoker"] = df["smoker"].fillna(
    df["smoker"].mode()[0]
)
df["region"] = df["region"].fillna(
    df["region"].mode()[0]
)
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=["charges"])
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
import pandas as pd

df_encode = pd.get_dummies(
    df,
    columns=["sex", "smoker", "region"],
    dtype=int
)
df_encode.head()
```

```python
X_df=df_encode.drop(
    'charges',
    axis=1
)
y_df=df_encode['charges']


print("Input shape:",X_df.shape)
print("Target shape:",y_df.shape)
```

```python
from sklearn.model_selection import train_test_split




X_train_df, X_test_df, y_train_df, y_test_df = train_test_split(
    X_df,
    y_df,
    train_size=0.8,
    test_size=0.20,
    random_state=42
)


print("Training rows:",X_train_df.shape[0])
print("Test input shape:",X_test_df.shape[0])
```

```python
from sklearn.preprocessing import StandardScaler

df_scaler = StandardScaler()
scale_columns = ["age", "bmi", "children"]

X_train_df_scaled = X_train_df.copy()
X_test_df_scaled = X_test_df.copy()

X_train_df_scaled[scale_columns] = df_scaler.fit_transform(
    X_train_df[scale_columns]
)
X_test_df_scaled[scale_columns] = df_scaler.transform(
    X_test_df[scale_columns]
)
```

```python
X_train_df_scaled
```

```python
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()
```

```python
linear_model.fit(
    X_train_df_scaled,
    y_train_df
)
print("Linear Regression model trained successfully.")
```

```python
pred = linear_model.predict(
    X_test_df_scaled
)
```

```python
df_result = pd.DataFrame({
    "Actual Charges": y_test_df.values,
    "Predicted Charges": pred
})
df_result.head(3)
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

mae = mean_absolute_error(

    y_test_df,
    pred
)
mse = mean_squared_error(
    y_test_df,
    pred
)
r2 = r2_score(
    y_test_df,
    pred
)
rmse = mse ** 0.5
r2 = r2_score(
    y_test_df,
    pred
)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2:", r2)
```

```python
new_customer_encoded = pd.get_dummies(
    new_customer,
    columns=["sex", "smoker", "region"],
    dtype=int
)
new_customer_encoded = new_customer_encoded.reindex(
    columns=X_train_df_scaled.columns,
    fill_value=0
)
```
